# Module 0 - Le monde des agents IA

## 0.1 Qu'est-ce qu'un agent ?

On entend partout parler d'« agents IA », et le terme est presque toujours mal expliqué. Partons de la définition classique, celle de Russell et Norvig (*Artificial Intelligence: A Modern Approach*) :

> Un **agent** est une entité qui **perçoit** son environnement à l'aide de capteurs, et qui **agit** sur cet environnement à l'aide d'effecteurs, afin d'atteindre ses objectifs.

Dit plus simplement : **un agent, c'est une boucle.** Il part d'un objectif, perçoit l'état du monde (ici, grâce à des **outils**), décide d'une action, observe le résultat, et **recommence** tant que l'objectif n'est pas atteint.

<div align="center">
<img src="./boucle_agent.svg" alt="ouvrir dans VS Code pour voir le schéma" width="760"/>
</div>

Retenez trois conséquences, qui guideront tout le projet :

1. **Un agent n'est pas un LLM.** Un modèle de langage peut *aider* un agent (comprendre une demande en langage naturel, choisir un outil), mais ce qui fait l'agent, ce sont ses outils et sa capacité à se rattraper. D'ailleurs, les modules 1 à 9 de ce projet n'utilisent **aucun** LLM.
2. **Un script n'est pas un agent.** Un programme qui appelle trois fonctions à la suite est un script. Un programme qui constate que son plan dépasse le budget, choisit lui-même quoi sacrifier, recalcule, puis explique ce qu'il a fait : ça, c'est un agent.
3. **Un agent agit sur le monde.** Lire des données ne suffit pas. À la fin du socle, votre agent pourra écrire une réservation dans une base de données, et c'est précisément là que les garde-fous deviennent indispensables.

| | Script | Chatbot (LLM seul) | Agent |
|---|---|---|---|
| Poursuit un objectif | non : il déroule des instructions | non : il répond au dernier message | **oui** |
| Consulte le monde extérieur | parfois | non : il ne sait que ce qu'il a appris | **oui, grâce à ses outils** |
| Se corrige quand il échoue | non | non | **oui : c'est sa boucle** |
| Agit (écrit, réserve, envoie) | oui, aveuglément | non | **oui, sous garde-fous** |

## 0.2 Les cinq familles d'agents

La typologie classique distingue cinq familles, de la plus simple à la plus élaborée :

| Étiquette | Principe | Exemple |
|---|---|---|
| `reflexe_simple` | des règles *condition → action*, sans mémoire | une porte automatique |
| `reflexe_modele` | des règles, plus un **état interne** qui garde la trace de ce que l'agent ne perçoit plus | un régulateur de vitesse qui suit la voiture de devant, même quand elle sort un instant du champ du radar |
| `but` | cherche une suite d'actions qui mène à un **objectif** | un solveur de Sudoku |
| `utilite` | parmi les solutions qui atteignent l'objectif, choisit la **meilleure** selon une mesure de satisfaction | un programme d'échecs qui note chaque position |
| `apprenant` | améliore son comportement grâce à l'expérience | un système de recommandation qui apprend de vos clics |

Ces familles s'emboîtent : un agent à utilité a aussi un but, et un agent apprenant peut être n'importe lequel des quatre autres, doté d'un mécanisme d'apprentissage.

## 0.3 Spécifier un agent : le cadre PEAS

Avant de programmer un agent, on le **spécifie**. Le cadre PEAS impose de répondre à quatre questions :

- **P**erformance : à quoi mesure-t-on qu'il fait bien son travail ?
- **E**nvironnement : dans quel monde évolue-t-il ?
- **A**ctionneurs (ou effecteurs) : par quels moyens agit-il sur ce monde ?
- **S**enseurs (ou capteurs) : par quels moyens perçoit-il ce monde ?

## 0.4 Le paysage agentique en 2026

Votre agent sera volontairement « fait main », pour que vous en compreniez chaque rouage. Mais il faut savoir où il se situe dans l'écosystème actuel : les notions ci-dessous reviendront tout au long du projet.

**Les agents à base de LLM.** Depuis 2023, la plupart des agents dont on parle utilisent un grand modèle de langage comme « cerveau » : le LLM lit la demande, **choisit un outil** parmi ceux qu'on lui décrit, lit le résultat, et recommence jusqu'à pouvoir répondre. Le LLM n'exécute jamais rien lui-même : il *demande* un appel d'outil, et c'est le programme qui l'exécute. Ce mécanisme s'appelle l'**appel d'outils** (*tool calling*, ou *function calling*). Vous le construirez au Module 4, puis vous le brancherez sur un vrai LLM au Module 10.

**Les grands motifs de conception** (*design patterns*) :

| Motif | Idée | Dans ce projet |
|---|---|---|
| ReAct (*Reason + Act*) | alterner raisonnement, action (un outil) et observation | la boucle du Module 10 |
| Planifier puis corriger | établir un plan complet, puis le réviser tant qu'il ne satisfait pas les contraintes | la boucle du Module 5 |
| Réflexion, critique | l'agent, ou un second agent, relit et critique le résultat | extension C |
| Multi-agents | plusieurs agents spécialisés, coordonnés par un orchestrateur | extension C |
| Humain dans la boucle | l'agent propose, l'humain valide les actions importantes | Module 8 |
| Mémoire | l'agent garde des informations d'une session à l'autre | extension B |

**Les protocoles.** Pour qu'un agent puisse utiliser des outils écrits par d'autres, des standards ont émergé : le **Model Context Protocol (MCP)**, proposé par Anthropic fin 2024, décrit comment exposer des outils et des données à n'importe quel agent compatible ; le protocole **Agent2Agent (A2A)**, proposé par Google en 2025, permet à des agents de dialoguer entre eux.

**Les frameworks.** LangGraph, CrewAI, LlamaIndex, smolagents (Hugging Face), Microsoft Agent Framework (héritier d'AutoGen), OpenAI Agents SDK, Claude Agent SDK... Ils fournissent la boucle, la gestion des outils et la mémoire. Ils sont précieux en production, mais ils masquent exactement ce que ce projet veut vous faire comprendre. Le paysage évolue très vite : considérez cette liste comme un instantané, pas comme une référence.

**Les risques.** Un agent qui agit peut se tromper *en agissant* : réserver le mauvais vol, tourner en boucle indéfiniment, inventer une solution qui n'existe pas (hallucination), obéir à une instruction malveillante cachée dans une donnée (*injection de prompt*), coûter cher en appels d'API. Chacun de ces risques a sa parade, et vous en construirez plusieurs : garde-fou anti-boucle (Module 5), honnêteté (Module 6), invariants testés (Module 7), confirmation humaine (Module 8), moindre privilège (Module 10).

# Module 2 - Les outils SQL

Un **outil**, pour un agent, c'est une fonction qui va chercher un fait précis dans le monde extérieur. Le premier, `outil_vols`, interroge la base de l'aéroport et répond à la question : « quels avions partent vers cette ville, ce jour-là, et à quel prix ? ».

### Une règle de sécurité : les requêtes paramétrées

Jusqu'ici, vous avez sans doute écrit vos requêtes avec les valeurs directement dedans : `WHERE origine = 'Nice'`. Dans un outil, c'est différent : la destination et la date arrivent en **paramètres de la fonction**, et elles changent à chaque appel. La tentation naturelle serait de fabriquer la requête avec un f-string :

```python
sql = f"SELECT * FROM vols WHERE destination = '{destination}'"   # surtout pas !
```

Ça semble fonctionner... jusqu'au jour où ça casse. D'abord, si la valeur contient une apostrophe (une ville qui s'appellerait `L'Escala`), l'apostrophe ferme la chaîne SQL au mauvais endroit et la requête devient invalide. Ensuite, et c'est bien plus grave : si la valeur vient d'un utilisateur (ou d'un LLM, au Module 10 !), elle peut contenir du code SQL qui sera exécuté tel quel. C'est une **injection SQL**, l'une des failles de sécurité les plus exploitées du web depuis 25 ans.

La solution standard est simple : on écrit un point d'interrogation `?` à la place de chaque valeur, et on fournit les valeurs à part, dans un tuple. La base remplace elle-même chaque `?` par la valeur correspondante, dans l'ordre, sans jamais l'interpréter comme du code :

```python
requete(conn, "SELECT * FROM vols WHERE destination = ? AND date_depart = ?", ("Madrid", "2026-08-12"))
```

### Rappel : la recherche par le sens

Un modèle d'embeddings transforme un texte en un vecteur de nombres, de telle sorte que deux textes qui parlent de la même chose, même avec des mots différents, donnent des vecteurs **proches**. Pour mesurer cette proximité, on utilise la **similarité cosinus** : le cosinus de l'angle entre les deux vecteurs, qui vaut 1 s'ils pointent dans la même direction.

# Module 4 - Décrire ses outils : le registre (*tool calling*)

Nos trois outils sont des fonctions Python. Pour **nous**, c'est suffisant : nous savons ce qu'elles font, et nous les appellerons nous-mêmes dans la boucle (Module 5). Mais imaginez qu'un **LLM** doive décider seul quel outil appeler, et avec quels arguments : c'est ce qui se passera au Module 10. Il ne voit pas votre code. Il ne voit qu'une **fiche descriptive** de chaque outil :

- un **nom** (`chercher_vols`) ;
- une **description** en langage naturel : ce que fait l'outil, ce qu'il retourne, *quand* l'utiliser ;
- un **schéma des paramètres**, au format *JSON Schema* : leurs noms, leurs types, lesquels sont obligatoires, les valeurs permises.

Quand le LLM veut utiliser un outil, il produit une **demande d'appel**, par exemple `{"name": "chercher_vols", "input": {"destination": "Madrid", "date_depart": "2026-08-12"}}`. Il n'exécute rien. C'est votre programme qui reçoit cette demande, appelle la vraie fonction et lui renvoie le résultat. Cette partie du programme s'appelle un **répartiteur** (*dispatcher*).

Deux leçons importantes :

1. **La description est un prompt.** Le LLM choisit ses outils uniquement d'après leurs fiches. Une description vague donne des appels au hasard. Une liste `enum` des valeurs permises l'empêche d'inventer une ville qui n'existe pas.
2. **Le répartiteur ne doit jamais planter.** Un LLM peut demander un outil qui n'existe pas, oublier un argument, se tromper de type. Si votre programme lève une exception, l'agent meurt. S'il renvoie un **message d'erreur clair**, le LLM le lit... et se corrige. C'est encore la boucle de l'agent.

Les fiches ci-dessous suivent le format de l'API de Claude (`name`, `description`, `input_schema`). Les autres fournisseurs utilisent des formats presque identiques.

## Deux fonctions fournies

Avant d'attaquer le cœur du projet, il nous faut deux petites fonctions utilitaires. Ce ne sont pas des exercices, mais lisez-les : la seconde servira de fondation à quelque chose d'intéressant.

**`chiffrer_voyage`** calcule le prix d'un voyage avant qu'on le propose : le vol, plus l'hôtel multiplié par le nombre de nuits, plus les activités retenues. Une convention pour garder les calculs simples : **tout est compté par personne**, l'hôtel aussi. Le budget du voyageur est donc un budget par personne.

**`dates_voisines`** retourne les jours situés autour de la date demandée : la veille, le lendemain, puis l'avant-veille et le surlendemain. Pourquoi en aurions-nous besoin ? Patience, c'est le Module 6.

````python
from datetime import datetime, timedelta


def chiffrer_voyage(vol, hotel, activites, nuits):
    """
    Calcule le prix total d'un voyage, par personne.

    Arguments :
    vol -- une ligne du DataFrame des vols
    hotel -- une ligne du DataFrame des hôtels
    activites -- une liste de dictionnaires, chacun avec une clé "prix_eur"
    nuits -- le nombre de nuits

    Retourne :
    total -- le prix total en euros, par personne
    """
    prix_vol = vol["prix_eur"]
    prix_hotel = hotel["prix_nuit"] * nuits
    prix_activites = sum(a["prix_eur"] for a in activites)
    return float(prix_vol + prix_hotel + prix_activites)


def dates_voisines(date_depart, ecart=2):
    """
    Les jours à explorer autour de la date demandée, du plus proche au plus lointain.

    Arguments :
    date_depart -- le jour souhaité, au format "2026-08-12"
    ecart -- de combien de jours on accepte de s'éloigner

    Retourne :
    voisines -- la liste des autres dates, sans la date demandée elle-même
    """
    jour = datetime.strptime(date_depart, "%Y-%m-%d")
    voisines = []
    for decalage in range(1, ecart + 1):
        for signe in (-1, 1):
            voisines.append((jour + timedelta(days=signe * decalage)).strftime("%Y-%m-%d"))
    return voisines


vol_test = vols_madrid.iloc[0]
hotel_test = outil_hotels(brochures, encodeur, "Madrid", ENVIE).iloc[0]
activites_test = outil_activites(conn, "Madrid").to_dict("records")

print(f"Vol {vol_test['numero']} : {vol_test['prix_eur']:.0f} EUR")
print(f"{hotel_test['hotel']} x 4 nuits : {hotel_test['prix_nuit'] * 4:.0f} EUR")
print(f"{len(activites_test)} activités : {sum(a['prix_eur'] for a in activites_test):.0f} EUR")
print(f"TOTAL : {chiffrer_voyage(vol_test, hotel_test, activites_test, 4):.0f} EUR")
print(f"\nAutour du 12 août, l'agent pourra aussi regarder : {dates_voisines('2026-08-12')}")
````

## La boucle de l'agent

Nous y voilà : c'est le cœur du projet, la partie qui transforme trois outils en un véritable agent.

Le premier plan que l'agent essaie est toujours le même : le vol le moins cher, l'hôtel le plus pertinent, et toutes les activités de la ville. La plupart du temps, ce plan idéal dépasse le budget. Un simple script s'arrêterait là en répondant « impossible ». Votre agent, lui, va **relâcher une contrainte et réessayer**, autant de fois que nécessaire. Et chaque sacrifice est noté dans un **journal**, pour pouvoir tout expliquer à la fin.

Reste à décider ce qu'on sacrifie en premier, et cet ordre n'a rien d'anodin :

1. **D'abord les activités payantes**, en commençant par la plus chère. Notez bien le mot « payantes » : retirer une visite gratuite ne ferait rien économiser, ce serait du gâchis pur.
2. **Ensuite seulement l'hôtel**, en descendant vers des options moins chères. Pourquoi en dernier ? Parce que l'hôtel, le voyageur l'a explicitement décrit (« pour la famille, avec une piscine »). Sacrifier en premier ce qui a été demandé en premier, c'est la meilleure façon de le décevoir.
3. S'il n'y a plus rien à relâcher, l'agent **avoue** que c'est impossible. On y reviendra, c'est important.

Vous verrez aussi un `print` au milieu de la boucle. Il n'a aucun rôle dans le calcul : il affiche une ligne par combinaison essayée, pour que vous voyiez l'agent chercher au lieu de le deviner. Remarquez également le `for _ in range(30)` : c'est un **garde-fou anti-boucle infinie**. Aucun agent sérieux ne tourne sans limite.

Deux variables méritent qu'on s'y arrête avant d'écrire quoi que ce soit :

- **`ordre`** est la liste des hôtels dans l'ordre où on les essaiera. Elle contient des numéros de lignes du tableau `hotels` : d'abord `0`, l'hôtel le plus pertinent, puis uniquement ceux qui coûtent réellement moins cher que lui, rangés du plus cher au moins cher, pour descendre en douceur plutôt que de tomber d'un coup sur le moins cher de tous. Le `key=lambda j: hotels.iloc[j]["prix_nuit"]` signifie simplement « trier ces numéros selon le prix de l'hôtel correspondant ».
- **`position`** indique où l'on en est dans cette liste. Elle démarre à 0, et l'hôtel en cours d'essai est donc `hotels.iloc[ordre[position]]`.

La structure de la boucle est posée (`if`, `elif`, `else`), et la phrase du journal est déjà écrite pour chaque cas. Il vous reste cinq lignes.

**Premier cas, `if payantes:`** (il reste des activités payantes au programme ; la liste `payantes` est déjà calculée) :

1. `plus_chere` : la sortie la plus chère de `payantes`. Écrivez `max(payantes, key=lambda a: a["prix_eur"])` ;
2. `retenues` : la même liste, sans elle. Le plus propre est `[a for a in retenues if a is not plus_chere]` ;
3. `position` : remettez-la à 0, pour repartir du meilleur hôtel maintenant que le programme coûte moins cher.

**Deuxième cas, `elif position + 1 < len(ordre):`** (plus d'activité payante à sacrifier, mais il reste un hôtel moins cher à essayer) :

4. `position` : avancez d'un cran dans la liste `ordre`.

**Troisième cas, `else:`** (il n'y a plus rien à relâcher) :

5. `impossible` : passez-la à `True` ; la fonction s'arrêtera juste en dessous en retournant `None`.

### Exercice 5 - La boucle de décision

````python
def essayer_une_date(demande, conn, brochures, encodeur):
    """
    Compose le meilleur voyage possible pour UN jour de départ donné.

    Arguments :
    demande -- un dictionnaire avec destination, date_depart, nuits, voyageurs, budget_max, envie
    conn, brochures, encodeur -- les sources de données de l'agent

    Retourne :
    voyage -- un dictionnaire décrivant le voyage, ou None si rien ne rentre dans le budget
    journal -- la liste des ajustements faits par l'agent, une phrase par sacrifice
    """
    vols = outil_vols(conn, demande["destination"], demande["date_depart"], demande["voyageurs"])
    hotels = outil_hotels(brochures, encodeur, demande["destination"], demande["envie"])
    activites = outil_activites(conn, demande["destination"])

    journal = []
    if vols.empty or hotels.empty:
        return None, ["aucun vol ou aucun hôtel disponible pour cette destination et cette date"]

    vol = vols.iloc[0]                              # le vol le moins cher
    retenues = activites.to_dict("records")         # toutes les activités, pour commencer
    impossible = False

    # L'ordre dans lequel on essaiera les hôtels : le plus pertinent d'abord, puis ceux qui
    # coûtent vraiment moins cher que lui, du plus cher au moins cher, pour descendre en douceur.
    prix_du_premier = hotels.iloc[0]["prix_nuit"]
    replis = [j for j in range(1, len(hotels)) if hotels.iloc[j]["prix_nuit"] < prix_du_premier]
    ordre = [0] + sorted(replis, key=lambda j: hotels.iloc[j]["prix_nuit"], reverse=True)
    position = 0                                    # où on en est dans la liste ordre

    for _ in range(30):                             # garde-fou anti-boucle infinie
        hotel = hotels.iloc[ordre[position]]
        total = chiffrer_voyage(vol, hotel, retenues, demande["nuits"])

        # Une ligne par combinaison essayée, pour voir l'agent chercher.
        ecart = total - demande["budget_max"]
        verdict = "OK" if ecart <= 0 else f"dépasse de {ecart:.0f}"
        print(f"   {demande['date_depart'][8:]}/{demande['date_depart'][5:7]}  {hotel['hotel'][:20]:20s} "
              f"vol {vol['prix_eur']:4.0f} + hôtel {hotel['prix_nuit'] * demande['nuits']:5.0f} "
              f"+ sorties {sum(a['prix_eur'] for a in retenues):4.0f} = {total:6.0f} EUR   {verdict}")

        if total <= demande["budget_max"]:
            break                                   # le plan tient dans le budget : on s'arrête là

        payantes = [a for a in retenues if a["prix_eur"] > 0]

        if payantes:                                # repli n°1 : sacrifier une sortie
            ### START CODE HERE ###
            plus_chere = None       # (1) la sortie la plus chère
            retenues = None         # (2) on la retire du programme
            position = None         # (3) on repart du meilleur hôtel
            ### END CODE HERE ###
            journal.append(f"j'ai retiré « {plus_chere['nom']} » ({plus_chere['prix_eur']:.0f} EUR)")

        elif position + 1 < len(ordre):             # repli n°2 : descendre d'un hôtel
            ### START CODE HERE ###
            position = None         # (4) l'hôtel suivant de la liste
            ### END CODE HERE ###
            journal.append(f"{hotel['hotel']} restait trop cher, "
                           f"j'ai pris {hotels.iloc[ordre[position]]['hotel']} à la place")

        else:                                       # plus rien à relâcher
            ### START CODE HERE ###
            impossible = None       # (5) l'agent renonce
            ### END CODE HERE ###
            journal.append("je n'avais plus rien à sacrifier ce jour-là")

        if impossible:
            return None, journal

    voyage = {"destination": demande["destination"], "date_depart": demande["date_depart"],
              "nuits": demande["nuits"], "voyageurs": demande["voyageurs"],
              "vol": vol, "hotel": hotels.iloc[ordre[position]], "activites": retenues,
              "prix_total": total, "budget_max": demande["budget_max"]}
    return voyage, journal


demande = {"destination": "Barcelone", "date_depart": "2026-08-12", "nuits": 4, "voyageurs": 2,
           "budget_max": 750, "envie": ENVIE}

voyage, journal = essayer_une_date(demande, conn, brochures, encodeur)
print()
afficher_voyage(voyage, journal)

gratuites = [a["nom"] for a in outil_activites(conn, "Barcelone").to_dict("records")
             if a["prix_eur"] == 0]
gardees = [a["nom"] for a in voyage["activites"]]

assert voyage is not None, "Un voyage devrait être trouvé pour cette demande"
assert voyage["prix_total"] <= demande["budget_max"], "Le voyage retenu doit tenir dans le budget"
assert all(nom in gardees for nom in gratuites), \
    "Retirer une sortie gratuite ne fait rien économiser : l'agent ne doit sacrifier que du payant"

# Une deuxième demande, pour voir le repli n°2 : un palace avec un budget trop juste.
palace = dict(demande, envie="le grand luxe, un palace avec spa", budget_max=950)
voyage_palace, journal_palace = essayer_une_date(palace, conn, brochures, encodeur)
print()
afficher_voyage(voyage_palace, journal_palace)

assert any("à la place" in ligne for ligne in journal_palace), \
    "Sur cette demande, l'agent doit finir par changer d'hôtel"
print("\nExercice validé !")
````

Regardez bien la deuxième demande. Pour offrir un palace, l'agent a sacrifié **tout** le programme payant, puis il a changé d'hôtel. Le voyageur se retrouve avec un très bel hôtel... et plus rien à faire de la journée.

Est-ce le bon comportement ? C'est discutable, et c'est la conséquence directe de l'ordre choisi : les activités d'abord, l'hôtel en dernier. Un agent plus malin remarquerait qu'après avoir changé d'hôtel, le budget permet de remettre deux ou trois sorties au programme.

Retenez surtout ceci : **cet arbitrage n'est écrit nulle part dans les données, c'est nous qui l'avons défini.** Les cinq lignes que vous venez d'écrire encodent une opinion sur ce qui compte le plus dans un voyage, autrement dit la **fonction d'utilité** implicite de l'agent (Module 0). C'est le genre de choix qu'on vous demandera de défendre en réunion, et c'est très souvent la partie la plus importante du travail.

# Module 6 - La proactivité : chercher là où on ne vous a rien demandé

Votre agent sait composer le meilleur voyage pour **le jour qu'on lui a donné**. Un bon agent de voyage humain ferait autre chose : il regarderait aussi les jours d'à côté.

Et il aurait raison. Vous l'avez mesuré au Module 1 : le prix des vols varie d'un jour à l'autre, et pour une destination donnée les écarts peuvent être importants. Personne ne demande jamais « et si je partais un jour plus tôt ? », tout simplement parce qu'on ne sait pas que ça change quelque chose.

C'est exactement ce qu'on attend d'un agent : **qu'il aille chercher une information qu'on ne lui a pas demandée, et qu'il nous la signale.**

La mécanique est déjà écrite dans la cellule suivante :

1. l'agent compose le voyage pour la date demandée, comme au Module 5 ;
2. il recommence pour chaque jour voisin (grâce à `dates_voisines`) et garde ceux qui aboutissent, dans une liste `alternatives` ;
3. il compare, et si un jour voisin fait nettement mieux, il l'écrit dans le journal.

Un point de conception essentiel : **l'agent ne déplace jamais les dates tout seul.** Il rend le voyage demandé et signale l'occasion. Décider reste le rôle de l'utilisateur, exactement comme pour la réservation (Module 8).

Reste une question, la seule vraiment intéressante : **comment décider qu'un jour voisin est « meilleur » ?**

Le réflexe serait de comparer les prix. C'est un piège. L'agent s'arrête dès qu'il passe sous le budget : quel que soit le jour, il finit donc toujours *juste* sous la barre. Comparer les totaux, ce serait comparer des prix presque identiques et ne rien voir.

Ce qui change réellement d'un jour à l'autre, ce n'est pas ce que le voyage coûte, c'est **ce qu'il reste dedans**. Un vol moins cher, c'est une sortie de plus qu'on n'a pas eu à sacrifier. On compare donc d'abord le nombre d'activités conservées, et le prix seulement en cas d'égalité.

En Python, cela s'écrit avec une clé à deux niveaux : `key=lambda v: (len(v["activites"]), -v["prix_total"])`. Les tuples sont comparés élément par élément : le nombre de sorties prime, et le signe moins devant le prix signifie « à sorties égales, le moins cher gagne ».

### Exercice 6 - L'agent complet

Il vous reste trois lignes :

1. `meilleure` : parmi la liste `alternatives`, le voyage qui gagne selon cette clé. Attention : c'est un `max`, pas un `min` ;
2. `sorties_en_plus` : combien d'activités `meilleure` garde de plus que `voyage` ;
3. `economie` : combien d'euros `meilleure` coûte de moins que `voyage`.

````python
def planifier(demande, conn, brochures, encodeur):
    """
    L'agent complet : il compose le voyage demandé, puis explore les jours voisins.

    Arguments :
    demande -- un dictionnaire avec destination, date_depart, nuits, voyageurs, budget_max, envie
    conn, brochures, encodeur -- les sources de données de l'agent

    Retourne :
    voyage -- le voyage pour la date demandée, ou None si rien n'y rentre
    journal -- les ajustements de l'agent, et le bon plan qu'il a repéré ailleurs
    """
    print(f"[agent] {demande['destination']}, {demande['nuits']} nuits, "
          f"{demande['voyageurs']} voyageur(s), budget {demande['budget_max']:.0f} EUR par personne")

    voyage, journal = essayer_une_date(demande, conn, brochures, encodeur)

    # On rejoue exactement le même raisonnement pour chaque jour voisin. Seul le prix du
    # vol change d'un jour à l'autre : les hôtels et les activités, eux, ne bougent pas.
    alternatives = []
    for autre_jour in dates_voisines(demande["date_depart"]):
        candidat, _ = essayer_une_date(dict(demande, date_depart=autre_jour), conn, brochures, encodeur)
        if candidat is not None:
            alternatives.append(candidat)

    if not alternatives:
        return voyage, journal

    if voyage is None:
        # Rien ne rentrait le jour demandé : on signale le jour voisin le moins cher.
        secours = min(alternatives, key=lambda v: v["prix_total"])
        jour = datetime.strptime(secours["date_depart"], "%Y-%m-%d").strftime("%d/%m")
        journal.append(f"en revanche, en partant le {jour}, un voyage à "
                       f"{secours['prix_total']:.0f} EUR par personne devenait possible")
        return voyage, journal

    ### START CODE HERE ###

    meilleure = None          # (1) le meilleur jour voisin
    sorties_en_plus = None    # (2) les sorties gagnées
    economie = None           # (3) les euros gagnés

    ### END CODE HERE ###

    jour = datetime.strptime(meilleure["date_depart"], "%Y-%m-%d").strftime("%d/%m")
    if sorties_en_plus > 0:
        journal.append(f"au passage, en partant le {jour} vous gardiez "
                       f"{len(meilleure['activites'])} sorties au lieu de {len(voyage['activites'])}, "
                       f"pour {meilleure['prix_total']:.0f} EUR")
    elif economie >= 10:
        journal.append(f"au passage, en partant le {jour} le même programme revenait à "
                       f"{meilleure['prix_total']:.0f} EUR, soit {economie:.0f} EUR de moins par personne")

    return voyage, journal


voyage, journal = planifier(demande, conn, brochures, encodeur)
print()
afficher_voyage(voyage, journal)

print()
print(texte_proposition(voyage, journal))

assert voyage is not None, "Un voyage devrait être trouvé pour cette demande"
assert voyage["date_depart"] == demande["date_depart"], \
    "L'agent ne doit jamais déplacer les dates tout seul : il signale, il ne décide pas"
assert any("en partant le" in ligne for ligne in journal), \
    "L'agent devrait avoir repéré un jour voisin plus avantageux et l'avoir noté dans le journal"
print("\nExercice validé !")
````

Relisez le journal affiché au-dessus, c'est lui la vraie nouveauté. L'agent a essayé, constaté que ça dépassait, choisi quoi sacrifier, recommencé, exploré des jours que personne ne lui avait demandé d'explorer, et il est capable de dire exactement ce qu'il a fait et pourquoi.

Et remarquez qu'à aucun moment un modèle de langage n'est intervenu. Le texte de la proposition est produit par votre code, à partir des chiffres calculés : il est donc toujours **exact**. Un modèle génératif placé ici n'aurait rien apporté, si ce n'est le risque de raconter joliment une décision qu'il n'a pas prise.

Il reste deux cas à examiner. Le premier : la date demandée ne passe pas, mais un jour voisin, si.

````python
serre = dict(demande, budget_max=660)   # trop juste pour le 12 août, mais pas pour tous les jours
voyage_serre, journal_serre = planifier(serre, conn, brochures, encodeur)
print()
afficher_voyage(voyage_serre, journal_serre)

assert voyage_serre is None, "Avec ce budget, rien ne doit rentrer le jour demandé"
assert any("devenait possible" in ligne for ligne in journal_serre), \
    "L'agent devrait avoir trouvé un jour voisin où le voyage tient"
print("\nPersonne ne lui avait demandé de regarder les autres jours. C'est ça, un agent.")
````

Voilà la phrase que votre code vient de produire tout seul : *« aucun voyage possible le jour demandé, mais en partant un autre jour, oui »*. Le voyageur n'avait pas posé la question. Il ne savait même pas qu'il y avait une question à poser.

Retenez le raisonnement, il est réutilisable partout : **quand une recherche échoue, l'élargir d'un cran coûte presque rien et rapporte beaucoup.** Ici, ça a coûté quelques requêtes SQL de plus.

Dernier cas : celui où il n'y a vraiment aucune solution, nulle part.

````python
impossible = dict(demande, budget_max=200)
voyage_impossible, journal_impossible = planifier(impossible, conn, brochures, encodeur)
print()
afficher_voyage(voyage_impossible, journal_impossible)

assert voyage_impossible is None, "Avec 200 EUR, l'agent ne doit rien proposer du tout"
assert not any("devenait possible" in l for l in journal_impossible), \
    "Aucun jour ne tient dans 200 EUR : l'agent ne doit rien promettre"
print("\nUn agent qui invente une solution quand il n'y en a pas est un agent dangereux. Celui-ci sait dire non.")
````

# Module 7 - Évaluer un agent

Jusqu'ici, chaque exercice a été validé sur **un ou deux exemples**. C'est suffisant pour vérifier qu'une fonction marche, pas pour affirmer qu'un agent est fiable. En pratique, on évalue un agent sur une **batterie de scénarios**, un *banc d'essai* (ou *benchmark*), et l'on mesure des indicateurs agrégés.

Une difficulté propre aux agents : il n'existe souvent **pas de réponse exacte attendue**. Quel est « le bon voyage » pour Rome avec 600 euros ? Personne ne le sait. En revanche, on sait énoncer des **invariants** : des propriétés qui doivent être vraies pour *toutes* les réponses.

| Invariant | Ce qu'il protège | Seuil exigé |
|---|---|---|
| un voyage proposé tient dans le budget | la promesse faite au client | 100 % |
| l'agent ne déplace jamais les dates de lui-même | l'autonomie de l'utilisateur | 100 % |
| quand rien n'est possible, l'agent ne propose rien | l'honnêteté | 100 % |

Et des **indicateurs de qualité**, qu'on cherche à améliorer sans seuil absolu : le taux de succès, le nombre moyen d'activités conservées, le nombre de lignes du journal, le temps de réponse...

### Exercice 7 - Le banc d'essai

La liste `SCENARIOS` contient huit demandes variées : des villes, des durées et des budgets différents, un départ le tout premier jour de la saison (il n'y a pas de vols la veille), la demande trop serrée du Module 6, et une demande impossible. La fonction `evaluer_agent` fait tourner votre agent sur chacune, en coupant ses `print` grâce à `contextlib.redirect_stdout`, et chronomètre chaque appel. À vous de calculer, pour chaque scénario, les quatre mesures de la zone d'exercice :

1. `trouve` : l'agent a-t-il proposé un voyage ? (`True` ou `False`) ;
2. `dans_budget` : le voyage proposé tient-il dans le budget ? (`False` si rien n'a été trouvé) ;
3. `dates_respectees` : la date du voyage proposé est-elle celle demandée ? (`True` si rien n'a été trouvé : il n'y a rien à reprocher) ;
4. `nb_activites` : combien d'activités le voyage conserve-t-il ? (0 si rien n'a été trouvé).

````python
SCENARIOS = [
    {"destination": "Barcelone", "date_depart": "2026-08-12", "nuits": 4, "voyageurs": 2, "budget_max": 750,
     "envie": ENVIE},
    {"destination": "Madrid", "date_depart": "2026-08-12", "nuits": 4, "voyageurs": 2, "budget_max": 1000,
     "envie": "des musées et une ambiance culturelle"},
    {"destination": "Rome", "date_depart": "2026-07-15", "nuits": 3, "voyageurs": 1, "budget_max": 600,
     "envie": "un hôtel calme et romantique"},
    {"destination": "Lisbonne", "date_depart": "2026-08-20", "nuits": 5, "voyageurs": 2, "budget_max": 900,
     "envie": "faire la fête et sortir le soir"},
    {"destination": "Vienne", "date_depart": "2026-08-01", "nuits": 3, "voyageurs": 4, "budget_max": 500,
     "envie": "le grand luxe, un palace avec spa"},
    {"destination": "Porto", "date_depart": "2026-07-01", "nuits": 3, "voyageurs": 2, "budget_max": 700,
     "envie": ENVIE},
    {"destination": "Barcelone", "date_depart": "2026-08-12", "nuits": 4, "voyageurs": 2, "budget_max": 660,
     "envie": ENVIE},
    {"destination": "Prague", "date_depart": "2026-07-20", "nuits": 4, "voyageurs": 2, "budget_max": 200,
     "envie": "un hôtel calme"},
]


def evaluer_agent(scenarios):
    """
    Fait tourner l'agent sur chaque scénario et mesure son comportement.

    Arguments :
    scenarios -- une liste de demandes

    Retourne :
    resultats -- un DataFrame, une ligne par scénario
    """
    lignes = []
    for demande in scenarios:
        debut = time.perf_counter()
        with contextlib.redirect_stdout(io.StringIO()):     # on fait taire les print de l'agent
            voyage, journal = planifier(demande, conn, brochures, encodeur)
        duree = time.perf_counter() - debut

        ### START CODE HERE ###

        trouve = None               # (1)
        dans_budget = None          # (2)
        dates_respectees = None     # (3)
        nb_activites = None         # (4)

        ### END CODE HERE ###

        lignes.append({"destination": demande["destination"], "date": demande["date_depart"],
                       "budget": demande["budget_max"], "trouve": trouve,
                       "prix_total": round(voyage["prix_total"]) if voyage else None,
                       "dans_budget": dans_budget, "dates_respectees": dates_respectees,
                       "nb_activites": nb_activites, "lignes_journal": len(journal),
                       "duree_s": round(duree, 2)})
    return pd.DataFrame(lignes)


resultats = evaluer_agent(SCENARIOS)
display(resultats)

trouves = resultats[resultats["trouve"] == True]
indicateurs = pd.Series({
    "taux de succès": resultats["trouve"].mean(),
    "respect du budget (voyages proposés)": trouves["dans_budget"].mean(),
    "respect des dates": resultats["dates_respectees"].mean(),
    "activités conservées (moyenne)": trouves["nb_activites"].mean(),
    "lignes de journal (moyenne)": resultats["lignes_journal"].mean(),
    "temps de réponse moyen (s)": resultats["duree_s"].mean(),
}).round(2)
display(indicateurs.to_frame("valeur"))

assert len(resultats) == len(SCENARIOS), "Une ligne par scénario"
assert resultats["trouve"].dtype == bool, "trouve doit valoir True ou False"
assert resultats["trouve"].sum() >= 4, "Votre agent devrait trouver un voyage pour au moins la moitié des scénarios"
assert not resultats.loc[resultats["destination"] == "Prague", "trouve"].any(), \
    "Invariant d'honnêteté violé : aucun voyage n'est possible à Prague avec 200 EUR"
assert indicateurs["respect du budget (voyages proposés)"] == 1, "Invariant violé : un voyage proposé dépasse le budget"
assert indicateurs["respect des dates"] == 1, "Invariant violé : l'agent a déplacé des dates"
assert (resultats.loc[~resultats["trouve"], "nb_activites"] == 0).all(), \
    "Quand rien n'est trouvé, nb_activites doit valoir 0"
print("Exercice validé ! Les trois invariants tiennent sur tout le banc d'essai.")
````

Lisez le tableau ligne par ligne : chaque ligne est un petit récit. Et regardez la colonne `duree_s` : chaque appel à l'agent prend un temps non négligeable. La raison : `planifier` appelle `essayer_une_date` cinq fois (le jour demandé et quatre voisins), et chaque appel **ré-encode** les brochures de la ville avec le modèle d'embeddings, alors qu'elles ne changent jamais. Pré-calculer ces vecteurs une fois pour toutes diviserait le temps de réponse : c'est l'extension H. Voilà à quoi sert un banc d'essai : il transforme une impression (« c'est un peu lent ») en un chiffre qu'on peut améliorer.

# Module 8 - Agir avec des garde-fous : la réservation

Jusqu'ici, votre agent n'a fait que **lire** et **analyser** des données. Il est temps de le laisser **agir** : écrire une réservation dans la table `reservations` de la base.

C'est un changement de nature. Une lecture ratée donne une mauvaise réponse ; une écriture ratée crée un **fait** dans le monde : un vol réservé au mauvais nom, une double facturation. C'est pourquoi toute action d'un agent qui modifie le monde doit être encadrée. Le premier garde-fou, et le plus important, s'appelle **l'humain dans la boucle** (*human-in-the-loop*) : l'agent propose, l'humain confirme.

Dans la cellule suivante, **lisez d'abord le garde-fou** qui est écrit pour vous : si `confirme` vaut `False`, la fonction ressort immédiatement en expliquant ce qu'elle *aurait* réservé, sans toucher à la base. L'utilisateur garde ainsi la main sur ce que fait l'agent.

### Exercice 8 - Réserver, mais seulement sur confirmation

# Module 10 (avancé) - Brancher un LLM : l'agent piloté par Claude

Votre agent a une faiblesse évidente : il faut remplir des champs. Un utilisateur, lui, écrit : *« On est deux, on voudrait aller à Barcelone du 12 au 16 août avec 750 euros chacun, un hôtel familial avec piscine »*. Transformer cette phrase en demande structurée, poser une question quand une information manque, choisir quel outil appeler, rédiger la réponse : c'est exactement ce que les LLM font bien.

L'architecture que nous allons construire est **hybride**, et c'est la leçon principale de ce module :

- le **LLM** s'occupe du **langage** : comprendre la demande, choisir les outils, rédiger la réponse ;
- vos **outils déterministes** s'occupent des **faits et des calculs** : les prix viennent de la base, et les additions et les arbitrages sont faits par `planifier`. Le LLM n'a pas le droit de faire les additions lui-même : un LLM qui calcule un total de tête peut se tromper, et avec aplomb ;
- l'**action irréversible**, `reserver`, n'est **pas** dans les outils du LLM : c'est le principe du **moindre privilège**. Le LLM propose, l'humain confirme dans l'application.

## 10.1 Un quatrième outil : l'agent lui-même

La boucle d'un agent LLM est la même que celle du Module 5, seul le décideur change :

1. on envoie au LLM la conversation et le **registre d'outils** (Module 4) ;
2. s'il répond par du texte (`stop_reason == "end_turn"`), c'est fini ;
3. s'il demande un ou plusieurs outils (`stop_reason == "tool_use"`), on ajoute sa réponse telle quelle à la conversation, on exécute chaque outil avec le **répartiteur** (Module 4), et on renvoie **tous** les résultats dans **un seul** message utilisateur, chacun dans un bloc `tool_result` qui rappelle l'identifiant (`tool_use_id`) de la demande à laquelle il répond ;
4. on recommence, avec un **garde-fou** sur le nombre de tours.

Commençons par donner au LLM un quatrième outil, `planifier_voyage`, qui enveloppe votre agent complet. On obtient une hiérarchie : **un agent LLM qui délègue à un agent déterministe**. La fonction `resumer_voyage` transforme le voyage (qui contient des lignes pandas) en un dictionnaire simple, lisible par le LLM. Rien à écrire dans cette cellule : lisez-la, en particulier les consignes données au modèle.

````python
MODELE = "claude-opus-5"

CONSIGNES = """Tu es un agent de voyage. Tu aides l'utilisateur à composer un voyage (vol, hôtel, activités) \
à l'aide des outils fournis.

Règles :
- Tu ne connais rien par toi-même : chaque prix, vol, hôtel ou activité que tu mentionnes doit venir d'un résultat d'outil.
- Pour proposer un voyage complet, appelle planifier_voyage : c'est lui qui fait les calculs et les arbitrages. \
Ne refais jamais les additions toi-même.
- Si une information indispensable manque (destination, dates, nombre de voyageurs, budget), demande-la au lieu de l'inventer.
- Les départs possibles vont du 2026-07-01 au 2026-08-31. Le budget s'entend par personne, tout compris.
- Tu ne peux pas réserver. Si l'utilisateur veut réserver, explique-lui qu'il doit confirmer dans l'application.
- Réponds en français, en quelques phrases, en reprenant fidèlement le journal de l'agent : ce qui a été sacrifié, \
et les jours voisins intéressants."""


def resumer_voyage(voyage, journal):
    """Transforme le voyage retourné par planifier en un dictionnaire simple, lisible par un LLM."""
    if voyage is None:
        return {"voyage": None, "journal": journal}
    v, h = voyage["vol"], voyage["hotel"]
    return {
        "destination": voyage["destination"], "date_depart": voyage["date_depart"],
        "nuits": voyage["nuits"], "voyageurs": voyage["voyageurs"],
        "vol": {"numero": v["numero"], "origine": v["origine"], "heure_depart": v["heure_depart"],
                "prix_eur": float(v["prix_eur"])},
        "hotel": {"nom": h["hotel"], "etoiles": int(h["etoiles"]), "note": float(h["note"]),
                  "prix_nuit": float(h["prix_nuit"])},
        "activites": [a["nom"] for a in voyage["activites"]],
        "prix_total_par_personne": round(voyage["prix_total"], 2),
        "budget_max": voyage["budget_max"],
        "journal": journal,
    }


def outil_planifier(arguments):
    """L'agent complet, emballé comme un outil : il reçoit les arguments du LLM, et se tait."""
    demande = {cle: arguments[cle] for cle in
               ("destination", "date_depart", "nuits", "voyageurs", "budget_max", "envie")}
    with contextlib.redirect_stdout(io.StringIO()):
        voyage, journal = planifier(demande, conn, brochures, encodeur)
    return resumer_voyage(voyage, journal)


SCHEMA_PLANIFIER = {
    "name": "planifier_voyage",
    "description": ("Compose le meilleur voyage complet (vol le moins cher, hôtel le plus pertinent, activités) "
                    "en respectant le budget par personne. Si le budget ne suffit pas, sacrifie d'abord des "
                    "activités payantes, puis descend en gamme d'hôtel, et signale les jours de départ voisins "
                    "plus avantageux. Retourne le voyage (ou null si rien n'est possible) et le journal des "
                    "ajustements. C'est le seul outil qui calcule des prix : utilise-le pour toute proposition."),
    "input_schema": {
        "type": "object",
        "properties": {
            "destination": {"type": "string", "enum": VILLES},
            "date_depart": {"type": "string", "description": "Le jour du départ, AAAA-MM-JJ, entre 2026-07-01 et 2026-08-31"},
            "nuits": {"type": "integer", "minimum": 1, "description": "Le nombre de nuits sur place"},
            "voyageurs": {"type": "integer", "minimum": 1},
            "budget_max": {"type": "number", "description": "Le budget maximal PAR PERSONNE, tout compris, en euros"},
            "envie": {"type": "string", "description": "Les envies du voyageur pour l'hôtel et le séjour, en texte libre"},
        },
        "required": ["destination", "date_depart", "nuits", "voyageurs", "budget_max", "envie"],
    },
}

FONCTIONS_OUTILS["planifier_voyage"] = outil_planifier
OUTILS_LLM = OUTILS + [SCHEMA_PLANIFIER]

print("Outils visibles par le LLM :", [o["name"] for o in OUTILS_LLM])
print("\nCe que le LLM lirait après un appel à planifier_voyage :")
print(en_json(executer_outil("planifier_voyage", {"destination": "Barcelone", "date_depart": "2026-08-12",
                                                  "nuits": 4, "voyageurs": 2, "budget_max": 750,
                                                  "envie": ENVIE})))
````

## 10.2 La boucle de l'agent LLM

Voici la boucle complète, à trois lignes près. Pour la tester **sans clé d'API et sans réseau**, nous utilisons `ClientSimule` (dans `utils.py`) : un faux client qui rejoue un scénario écrit à l'avance, et qui **vérifie** à chaque tour que vous lui renvoyez la conversation au bon format, comme le ferait la vraie API. Tester un agent LLM avec un LLM simulé est une pratique courante : c'est rapide, gratuit, reproductible, et ça isole les bugs de *votre* boucle de ceux du modèle.

Le scénario simulé met votre boucle à l'épreuve : au premier tour, le « modèle » demande deux outils **en même temps** ; au deuxième, il appelle l'agent complet ; au troisième, il demande un outil **qui n'existe pas** (`reserver`, que nous avons volontairement gardé hors de sa portée) ; au quatrième, il conclut.

### Exercice 10 - Les trois lignes de la boucle

Les formats à respecter :

```python
# (1) la réponse de l'assistant, ajoutée telle quelle à la conversation
{"role": "assistant", "content": reponse.content}

# (2) le résultat d'un outil
{"type": "tool_result", "tool_use_id": bloc.id, "content": en_json(resultat), "is_error": erreur}

# (3) tous les résultats du tour, dans UN SEUL message utilisateur
{"role": "user", "content": resultats}
```

Deux détails de l'appel à l'API méritent une explication. Le paramètre `fallbacks="default"` (avec l'en-tête bêta qui l'accompagne) demande à l'API, si le modèle refuse une demande pour des raisons de sécurité, de la relancer automatiquement sur un autre modèle. Et `max_tokens` borne la longueur de chaque réponse : c'est un garde-fou de coût.

````python
def agent_llm(question, client, max_tours=8):
    """
    Un agent LLM : le modèle choisit les outils, notre code les exécute, jusqu'à la réponse finale.

    Arguments :
    question -- la demande de l'utilisateur, en langage naturel
    client -- un client Anthropic (ou un ClientSimule, pour les tests)
    max_tours -- le garde-fou : nombre maximal d'allers-retours avec le modèle

    Retourne :
    reponse -- le texte final du modèle
    """
    messages = [{"role": "user", "content": question}]

    for tour in range(1, max_tours + 1):
        reponse = client.beta.messages.create(
            model=MODELE,
            max_tokens=16000,
            system=CONSIGNES,
            tools=OUTILS_LLM,
            messages=messages,
            betas=["server-side-fallback-2026-07-01"],
            fallbacks="default",
        )

        if reponse.stop_reason != "tool_use":       # le modèle a fini (ou refusé, ou manqué de place)
            texte = "\n".join(b.text for b in reponse.content if b.type == "text")
            return texte or f"(le modèle s'est arrêté sans texte : {reponse.stop_reason})"

        ### START CODE HERE ###
        messages.append(None)       # (1) la réponse de l'assistant, telle quelle
        ### END CODE HERE ###

        resultats = []
        for bloc in reponse.content:
            if bloc.type != "tool_use":
                continue
            print(f"  [tour {tour}] {bloc.name}({en_json(bloc.input)})")
            resultat = executer_outil(bloc.name, bloc.input)
            erreur = isinstance(resultat, dict) and "erreur" in resultat

            ### START CODE HERE ###
            resultats.append(None)  # (2) un bloc tool_result : type, tool_use_id, content, is_error
            ### END CODE HERE ###

        ### START CODE HERE ###
        messages.append(None)       # (3) UN SEUL message utilisateur avec tous les résultats
        ### END CODE HERE ###

    return "Je m'arrête : trop d'allers-retours sans conclusion (garde-fou max_tours)."


simule = ClientSimule([
    [("chercher_activites", {"ville": "Barcelone"}),                                   # tour 1 : deux outils
     ("chercher_hotels", {"ville": "Barcelone", "envie": "familial avec piscine"})],   # en même temps
    [("planifier_voyage", {"destination": "Barcelone", "date_depart": "2026-08-12", "nuits": 4,
                           "voyageurs": 2, "budget_max": 750, "envie": ENVIE})],       # tour 2 : l'agent complet
    [("reserver", {"client": "Marc"})],                                                # tour 3 : un outil interdit
    "Voici le voyage que je vous propose pour Barcelone...",                           # tour 4 : la réponse
])

reponse = agent_llm("On est deux, Barcelone du 12 au 16 août, 750 euros chacun, hôtel familial avec piscine.",
                    simule)
print("\nRéponse finale :", reponse)

assert simule.nb_appels == 4, f"Le modèle simulé attendait 4 appels, il en a reçu {simule.nb_appels}"
assert reponse.startswith("Voici"), "agent_llm doit retourner le texte final du modèle"
assert simule.erreurs_signalees == [False, False, False, True], \
    "Seul l'appel à l'outil inexistant « reserver » doit être signalé comme une erreur (is_error=True)"
assert '"prix_total_par_personne"' in simule.resultats_recus[2], \
    "Le résultat de planifier_voyage doit être transmis au modèle"

# Le garde-fou : un modèle qui demanderait des outils sans jamais conclure
bavard = ClientSimule([[("chercher_activites", {"ville": "Rome"})]] * 5)
reponse_bavard = agent_llm("Que faire à Rome ?", bavard, max_tours=3)
assert bavard.nb_appels == 3 and "garde-fou" in reponse_bavard, \
    "La boucle doit s'arrêter au bout de max_tours tours"
print("\nExercice validé ! Votre boucle parle le langage de l'API, et elle sait s'arrêter.")
````

# Bilan

Récapitulons ce que vous avez construit, concept par concept :

| Concept | Votre implémentation |
|---|---|
| Perception | trois outils : deux requêtes SQL paramétrées et une recherche sémantique |
| Description des outils | un registre au format *tool calling*, et un répartiteur qui ne plante jamais |
| Décision sous contrainte | une boucle qui chiffre, compare, sacrifie et recommence |
| Traçabilité | un journal qui explique chaque décision, sans rien inventer |
| Proactivité | l'exploration des jours voisins, sans jamais décider à la place de l'utilisateur |
| Honnêteté | un agent qui sait dire « impossible » |
| Évaluation | un banc d'essai, des invariants, des indicateurs |
| Action sous garde-fous | une réservation en deux temps, confirmée par un humain |
| Déploiement | une application web |
| Orchestration par un LLM | une boucle *tool use*, testée par simulation, avec le moindre privilège |

Remarquez que le LLM n'apparaît qu'à la toute fin de ce tableau. Ce n'est pas un hasard : **ce qui fait un bon agent, ce sont ses outils, sa boucle et ses garde-fous.** Un LLM rend l'agent plus agréable à utiliser et plus souple face au langage ; il ne le rend pas plus juste. Si vous ne deviez retenir qu'une idée de ce projet, ce serait celle-là.
